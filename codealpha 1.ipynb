```python
# ===================================
# CODEALPHA TASK 1 - WEB SCRAPING
# ===================================

# Import Libraries
import requests
from bs4 import BeautifulSoup
import pandas as pd

# -----------------------------------
# 1. Website URL
# -----------------------------------

url = "https://quotes.toscrape.com/"

# Send Request
response = requests.get(url)

# Parse HTML Content
soup = BeautifulSoup(response.text, 'html.parser')

# -----------------------------------
# 2. Extract Data
# -----------------------------------

quotes = []
authors = []

# Find all quote blocks
quote_blocks = soup.find_all('div', class_='quote')

# Extract quote text and author
for quote in quote_blocks:

    text = quote.find('span', class_='text').text
    author = quote.find('small', class_='author').text

    quotes.append(text)
    authors.append(author)

# -----------------------------------
# 3. Create Dataset
# -----------------------------------

df = pd.DataFrame({
    'Quote': quotes,
    'Author': authors
})

# Display Dataset
print("Scraped Dataset")
display(df)

# -----------------------------------
# 4. Save Dataset
# -----------------------------------

df.to_csv("scraped_quotes.csv", index=False)

print("\nDataset saved as scraped_quotes.csv")

# -----------------------------------
# 5. Insights
# -----------------------------------

print("\nInsights:")
print("1. Successfully extracted quotes and authors from website.")
print("2. HTML structure was analyzed using BeautifulSoup.")
print("3. Custom dataset was created for analysis.")
print("4. Web scraping helps collect real-world public data.")

# -----------------------------------
# 6. Final Conclusion
# -----------------------------------

print("\nConclusion:")
print("Web scraping successfully collected data from a public website.")
print("BeautifulSoup helped navigate HTML structure and extract accurate information.")
```
